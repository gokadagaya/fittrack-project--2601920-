Firstly i created a data which was provided and used a dictionary of list for cleanest approch.and printed the pandas dataframe as employee_data.

```python
import pandas as pd

data = {
    'EmployeeID': [201, 202, 203, 204, 205, 206, 207, 208, 209, 210],
    'Name': ['Sarah', 'Michael', 'Jennifer', 'David', 'Lisa',
             'James', 'Emma', 'Robert', 'Sophia', 'William'],
    'Department': ['Engineering', 'Sales', 'Marketing', 'Engineering', 'HR',
                   'Sales', 'Engineering', 'Marketing', 'Sales', 'HR'],
    'Salary': [85000, 62000, 71000, 92000, 58000,
               68000, 78000, 75000, 65000, 60000],
    'Performance Score': [4, 3, 5, 5, 4, 4, 3, 4, 5, 3],
    'Years of Service': [5, 3, 7, 8, 4, 6, 2, 5, 4, 3]
}
employee_data=pd.DataFrame(data)
employee_data
```

Select and display only the 'Name', 'Department', and 'Salary' columns

selecting only name , department and salary columns by the original dataframe employee_data

```python
selected_columns=employee_data[['Name','Department','Salary']]
selected_columns
```

Use iloc to display the employee at row position 5

taking original dataframe employee_data and using iloc(position based)


```python
employee=employee_data.iloc[5]
employee
```

Use loc to display the employee with EmployeeID 208

loc to filter the row where EmployeeID is 208 and returns only the Name and EmployeeID columns.

```python
emp=employee_data.loc[employee_data['EmployeeID']==208,['Name','EmployeeID']]
emp
```

Filter and display all employees from the 'Engineering' department

uses Pandas query() to filter and return only the rows where the Department column equals 'Engineering'.


```python
filtering=employee_data.query("Department =='Engineering'")
filtering
```

Filter and display all employees with Salary greater than 70000

uses Pandas query() to filter and return only the salary>7000

```python
employees=employee_data.query("Salary>70000")
employees
```

Filter and display employees from 'Sales' department with Performance Score >= 4


```python
employees=employee_data[(employee_data['Department']=='Sales')&(employee_data['Performance Score']>=4)]
employees
```

Sort the entire DataFrame by Salary in descending order and display the top 3 highest-paid employees


```python
highest_paid_employees = employee_data.sort_values(by='Salary', ascending=False).head(3)
highest_paid_employees
```

Sort by Department (ascending) and then by Salary (descending) within each department


```python
sorting=employee_data.sort_values(by=['Department','Salary'],ascending=[True,False])
sorting
```
